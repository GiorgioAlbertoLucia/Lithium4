```python
import sympy as sym
```

```python
sym.init_printing()
mh, mn, mp, M = sym.symbols('m_h m_n m_p M')
```

```python
matrix = sym.Matrix([[mh/M, mp/M, mn/M],
                     [0, -1, 1],
                     [1, -mp/(mp+mn), -mn/(mp+mn)]])
matrix
```

```python
sym.simplify(matrix.inv())
```

```python
matrix_inv = sym.simplify(matrix.inv())
matrix_equal = matrix_inv.subs({mh: mp, mn: mp, M: 3*mp})
matrix_equal
```

```python
matrix_inv = sym.simplify(matrix.inv())
matrix_equal = matrix_inv.subs({mn: mp})
matrix_equal
```
